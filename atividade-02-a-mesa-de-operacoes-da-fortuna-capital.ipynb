## 📖 Glossário de Termos Financeiros

Antes de começar, conheça os termos de mercado que aparecem nos
exercícios. Não é preciso decorá-los — volte aqui sempre que precisar.

### Mercado de ações (Exercícios 1, 3 e 4)

| Termo | O que é |
|---|---|
| **B3** | Bolsa de valores brasileira (Brasil, Bolsa, Balcão). É o ambiente onde ações, contratos futuros e outros ativos são negociados. |
| **Ticker** | Código que identifica um ativo na bolsa. Ex.: `PETR4` = Petrobras (ação preferencial). O número indica o tipo: 3 = ordinária (com voto), 4 = preferencial (com dividendo prioritário). |
| **Preço de fechamento** | Último preço pelo qual o ativo foi negociado no dia. |
| **Volume negociado** | Quantidade de ações que trocaram de mão durante o pregão (dia de negociação). |
| **Volume financeiro** | Preço × volume = quanto dinheiro circulou naquele ativo no dia. Mede a **liquidez** — ações com alto volume financeiro são fáceis de comprar e vender sem mexer muito no preço. |
| **Ordem** | Instrução de compra ou venda que o investidor envia à corretora. |
| **Corretagem** | Taxa cobrada pela corretora para executar a ordem na bolsa, como percentual do valor da operação. |
| **Pregão** | Sessão diária de negociação na B3. |
| **Back-office** | Setor operacional da corretora que confere e reconcilia as operações do dia após o fechamento do pregão. |
| **Preço médio** | Valor total da operação ÷ quantidade de ações. |

### Canais de negociação (Exercício 3)

| Canal | O que é |
|---|---|
| **Home broker** | Plataforma online onde o próprio investidor envia suas ordens. Taxa baixa, limite de valor por segurança. |
| **Assessor** | Profissional da corretora que executa a ordem para o cliente. Cobra mais, aceita ordens maiores. |
| **Algoritmo** (algotrading) | Software que envia ordens automaticamente. Taxa mínima. |
| **Mesa de operações** | Equipe especializada que executa ordens grandes ou complexas. Sem limite de valor, maior taxa. |

### Crédito corporativo (Exercício 2)

Os indicadores ILC, Margem EBITDA e Alavancagem estão explicados em detalhe
no enunciado do Exercício 2.

### Câmbio (Exercício 5)

| Termo | O que é |
|---|---|
| **Câmbio** | Troca de uma moeda por outra. |
| **Cotação** | Preço de uma moeda estrangeira em reais. |
| **Spread cambial** | Diferença entre a cotação de mercado e a que o banco/corretora oferece ao cliente — o "lucro" da instituição na operação. |
| **IOF** | Imposto sobre Operações Financeiras — tributo federal sobre câmbio, crédito, seguro e títulos. |
| **Alíquota** | Percentual do imposto aplicado sobre a base de cálculo. |

### Fundos de investimento (Exercício 6)

| Termo | O que é |
|---|---|
| **Fundo de investimento** | "Condomínio financeiro": investidores juntam dinheiro e um gestor profissional aplica segundo uma estratégia. |
| **CVM** | Comissão de Valores Mobiliários — regula e fiscaliza o mercado de capitais brasileiro. |
| **Renda Fixa** | Rendimento previsível (Tesouro Direto, CDBs, debêntures). |
| **Renda Variável** | Rendimento depende do mercado (ações, fundos imobiliários). |
| **Multimercado** | Investe em renda fixa, ações, câmbio e derivativos ao mesmo tempo. |
| **PL (Patrimônio Líquido)** | Valor total dos ativos do fundo. |
| **Alavancagem (em fundos)** | Fundo assume posições maiores que seu patrimônio, via derivativos ou empréstimos. |

### IPO e governança (Exercício 7)

| Termo | O que é |
|---|---|
| **IPO** | Oferta Pública Inicial — quando uma empresa vende ações na bolsa pela primeira vez. |
| **Margem líquida** | Lucro líquido ÷ receita líquida × 100. |
| **Governança corporativa** | Práticas que protegem os acionistas minoritários contra decisões abusivas dos controladores. |
| **Conselho independente** | Membros do Conselho de Administração que não são executivos nem acionistas. |
| **Auditoria Big Four** | Deloitte, PwC, EY ou KPMG — sinaliza transparência contábil. |
| **Tag along** | Direito do minoritário de vender suas ações pelo mesmo preço em caso de troca de controle. |

### Tributação de serviços (Exercício 8)

| Termo | O que é |
|---|---|
| **ISS** | Imposto Sobre Serviços — tributo municipal, alíquota de 2% a 5%. |
| **Simples Nacional** | Regime simplificado para micro e pequenas empresas, com alíquotas reduzidas. |
| **Lucro Presumido** | O governo "presume" uma margem de lucro e tributa sobre ela. |
| **Nota fiscal** | Documento que registra a prestação de um serviço e serve de base para os impostos.

## 🥉 Exercício 2 — O Classificador de Risco de Crédito: `if`/`elif`/`else` + operadores lógicos

A Fortuna Capital também opera crédito corporativo. Antes de aprovar uma
linha de crédito, o analista classifica o **grau de risco** da empresa a
partir de três indicadores.

**ILC — Índice de Liquidez Corrente** = Ativo Circulante ÷ Passivo
Circulante. Mede a capacidade de pagar dívidas de curto prazo. ILC ≥ 1,5 é
folga confortável; abaixo de 1,0 é sinal de alerta.

**Margem EBITDA** = EBITDA ÷ Receita Líquida × 100. Mostra quanto de cada
real faturado vira caixa operacional. ≥ 15% é eficiente; < 5% é pouco
lucrativa.

**ALAV — Alavancagem** = Dívida Líquida ÷ EBITDA. Em quantos anos de caixa a
empresa pagaria toda a dívida. ≤ 3,0 é pouco endividada; > 5,0 é risco alto.

### 📋 Regras de classificação (em cascata — a primeira verdadeira "ganha")

1. Se ILC ≥ 1,5 **e** ME ≥ 15% **e** ALAV ≤ 3,0 → **AAA** (risco mínimo)
2. Se ILC ≥ 1,0 **e** ME ≥ 10% **e** ALAV ≤ 5,0 → **BBB** (risco moderado)
3. Se ILC < 1,0 **ou** ME < 5% → **CCC** (risco alto)
4. Qualquer outro caso → **BB** (risco intermediário)

Decisão de crédito: **AAA/BBB** → ✅ aprovado · **BB** → ⚠️ garantia
adicional · **CCC** → ❌ negado.

> 💡 O **AAA** exige `and` (todos os três bons); o **CCC** usa `or` (basta
> **um** indicador ruim para reprovar). É a diferença entre critério de
> *aprovação* (precisa de todos) e de *exclusão* (basta um).

## 🥈 Exercício 4 — O Validador de Lançamentos: `try`/`except` + `raise`

O **back-office** reconcilia, depois do pregão, todos os lançamentos do
dia: cada operação de compra (`C`) ou venda (`V`) precisa ser validada, e o
**preço médio** (valor total ÷ quantidade) calculado.

Esses dados chegam **como texto** — `MEU_VALOR_LANCAMENTO`,
`MINHA_QUANTIDADE_LANCAMENTO`, `MEU_CODIGO_LANCAMENTO` — e estão sujeitos a
dois tipos de problema:

| Problema | Exemplo | Ferramenta |
|---|---|---|
| Dado ilegível | `"n/d"`, `""` | `try`/`except ValueError` |
| Regra de negócio violada | `"-5"`, código `"X"` | `raise ValueError` |

> 💡 Um único `except ValueError as erro` captura os dois tipos: o que o
